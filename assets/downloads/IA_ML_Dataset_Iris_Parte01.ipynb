## **Apresentação**

<img src="http://miro.medium.com/max/1000/1*Hh53mOF4Xy4eORjLilKOwA.png" width="700" height="200"/>

O Dataset da Flor de Íris ou o Dataset da Íris de Fisher é um conjunto de dados multivariados introduzido pelos britânico estatístico, eugenista e biólogo Ronald Fisher em seu artigo de 1936. O uso de múltiplas medições em problemas taxonômicos como um exemplo de análise discriminante linear. Às vezes é chamado de Dataset da Íris de Anderson, porque Edgar Anderson coletou os dados para quantificar a variação morfológica da flor de íris de três espécies relacionadas. Duas das três espécies foram coletadas na Península de Gaspé "todas da mesma pastagem, e colhidas no mesmo dia e medidas ao mesmo tempo pela mesma pessoa com o mesmo aparelho".


O Dataset consiste em 50 amostras de cada uma das três espécies de Íris (Íris Setosa , Íris Virginica e Íris Versicolor). Quatro características foram medidas em cada amostra: o comprimento e a largura das sépalas e pétalas, em centímetros. Com base na combinação dessas quatro características, Fisher desenvolveu um modelo discriminante linear para distinguir as espécies umas das outras.

## **Scikit-learn**

A scikit-learn (originalmente scikits.learn) é uma biblioteca de aprendizado de máquina de código aberto para a linguagem de programação Python. Ela inclui vários algoritmos de classificação, regressão e agrupamento incluindo máquinas de vetores de suporte, florestas aleatórias, gradient boosting, k-means e DBSCAN, e é projetada para interagir com as bibliotecas Python numéricas e científicas NumPy e SciPy.

Para conhecer mais, visite o site oficial: http://scikit-learn.org

## **Como ver/obter os dados do Dataset da Flor de Íris no Python?**

A biblioteca scikit-learn tem dentro de si, um grande conjunto de datasets para podermos utilizar.

Você pode, por exemplo, carregar somente o dataset da Flor de Íris para trabalhar:

```
from sklearn.datasets import load_iris
dados = load_iris()
```

Ou você, pode importar todos os datasets disponíveis e usar somente o que você quiser:

```
from sklearn import datasets
dados = datasets.load_iris()
```
